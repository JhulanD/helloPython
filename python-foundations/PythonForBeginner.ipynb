```python
calculation_of_units = 24 * 60 * 60
name_of_unit = 'seconds'

def days_to_units(num_of_days):
  print(f"{num_of_days} days in a month are {num_of_days * calculation_of_units} {name_of_unit}")

days_to_units(10)
days_to_units(30)
days_to_units(180)

# print(f"30 days in a month are {30 * calculation_of_units} {name_of_unit}")
# print(f"60 days in a month are {60 * calculation_of_units} {name_of_unit}")
# print(f"180 days in a month are {180 * calculation_of_units} {name_of_unit}")
```

```python
calculation_of_units = 24 * 60 * 60
name_of_unit = 'seconds'

def days_to_units(num_of_days, custom_msg):
  print(f"{num_of_days} days in a month are {num_of_days * calculation_of_units} {name_of_unit}")
  print(custom_msg)

days_to_units(10, 'Looks good!')
days_to_units(30, 'Looks better!' )
days_to_units(180, 'Looks awesome!')

# print(f"30 days in a month are {30 * calculation_of_units} {name_of_unit}")
# print(f"60 days in a month are {60 * calculation_of_units} {name_of_unit}")
# print(f"180 days in a month are {180 * calculation_of_units} {name_of_unit}")
```
