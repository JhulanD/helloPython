```python
df[df['Category'] == 'Electronics'].sort_values('Price', ascending = False)
# How to add multiple conditions and filters
# Solution: df[(df['Category'] == 'Electronics') & (df['Quantity'] > 2)]

df[(df['Category'] == 'Electronics') & (df['Country'] == 'USA')]
# Explanation: This is a boolean expression that checks if the category is Electronics and the country is USA

df[(df['Category'] == 'Electronics') | (df['Country'] == 'USA')]
# Explanation: This is a boolean expression that checks if the category is Electronics or the country is USA
```

```python
# When and where to use which brackets:-
# Solution - Use () for multiple conditions
# Solution - Use [] for single condition
# Solution - Use {} for multiple conditions
# Solution: https://www.geeksforgeeks.org/when-and-where-to-use-which-brackets-in-python/
# Examples: 
    # df[(df['Category'] == 'Electronics') & (df['Quantity'] > 2)] (means - and)
    # df[(df['Category'] == 'Electronics') & (df['Country'] == 'USA')] (Means - and)
    # df[(df['Category'] == 'Electronics') | (df['Country'] == 'USA')] (Means - or)
    # df[(df['Category'] == 'Electronics') ^ (df['Country'] == 'USA')] (Means - either or)
```

```python
df[(df['Category'] == 'Electronics') ^ (df['Country'] == 'USA')]
```

```python
df[df['CustomerName'].str.endswith('a') | df['CustomerName'].str.startswith('A')]
```

```python
df[~df['Country'].isin(['USA', 'Canada'])]
# Explain - This is a boolean expression that checks if the country is not USA or Canada and the sign is ~

df[df['Country'].isin(['USA', 'Canada'])]
# Explain - This is a boolean expression that checks if the country is USA or Canada 
```

```python
df.loc[df['Country'] == 'USA' , 'Country'] = 'United States'
df['Country'] = df['Country'].str.upper()
```
